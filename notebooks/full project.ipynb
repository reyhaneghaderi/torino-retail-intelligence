```python
# 5. Count somministrazione stores by zone
som_zone = (
    som_clean
    .groupby("statistical_zone_clean")
    .size()
    .reset_index(name="n_somministrazione")
)
som_zone 
```

```python
 #Merge with population
opportunity = som_zone.merge(
    pop[["statistical_zone_clean", "neighborhood_clean", "total_population"]],
    on="statistical_zone_clean",
    how="inner"
)

print("som zones:", len(som_zone))
print("population zones:", len(pop))
print("merged zones:", len(opportunity))
```

```python
#  Competition indicator
opportunity["stores_per_1000_residents"] = (
    opportunity["n_somministrazione"] / opportunity["total_population"]
) * 1000
```

```python
som_clean["x"] = (
    som_clean["x"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

som_clean["y"] = (
    som_clean["y"]
    .astype(str)
    .str.replace(",", ".", regex=False)
)

som_clean["x"] = pd.to_numeric(som_clean["x"], errors="coerce")
som_clean["y"] = pd.to_numeric(som_clean["y"], errors="coerce")
```

```python
som_clean[["x", "y"]].dtypes
```

```python
# Zone centers based on store coordinates
zone_centers = (
    som_clean
    .groupby("statistical_zone_clean")[["x", "y"]]
    .mean()
    .reset_index()
    .rename(columns={
        "x": "zone_center_x",
        "y": "zone_center_y"
    })
)
```

```python
#  Distance to nearest metro station


def nearest_metro_distance(row, metro_df):
    distances = np.sqrt(
        (metro_df["x"] - row["zone_center_x"])**2 +
        (metro_df["y"] - row["zone_center_y"])**2
    )
    return distances.min()

zone_centers["distance_to_nearest_metro"] = zone_centers.apply(
    lambda row: nearest_metro_distance(row, metro),
    axis=1
)
```

```python
#  Merge metro distance
opportunity = opportunity.merge(
    zone_centers[["statistical_zone_clean", "distance_to_nearest_metro"]],
    on="statistical_zone_clean",
    how="left"
)
```

```python
#  Normalize variables
def minmax(series):
    return (series - series.min()) / (series.max() - series.min())

opportunity["normalized_population"] = minmax(opportunity["total_population"])

opportunity["normalized_competition"] = minmax(
    opportunity["stores_per_1000_residents"]
)
opportunity["normalized_distance_to_metro"] = minmax(
    opportunity["distance_to_nearest_metro"]
)
```

```python
# Final opportunity score
opportunity["retail_opportunity_score"] = (
    0.45 * opportunity["normalized_population"]
    - 0.35 * opportunity["normalized_competition"]
    - 0.20 * opportunity["normalized_distance_to_metro"]
)
```

```python
#  Sort final result
opportunity_sorted = opportunity.sort_values(
    "retail_opportunity_score",
    ascending=False
)

opportunity_sorted.head(10)
```
